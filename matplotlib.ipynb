```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3]
y = [1, 4, 9]

plt.plot(x, y)
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Simple Line Graph")
plt.show()
```

```python
subjects = ['Maths', 'Physics', 'Chemistry']
marks = [85, 78, 90]

plt.bar(subjects, marks)
plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 7, 4, 8, 6]

plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
activities = ['Study', 'Sleep', 'Entertainment', 'Others']
hours = [6, 8, 4, 6]

plt.pie(hours, labels=activities, autopct='%1.1f%%')
plt.title("Daily Activities")
plt.show()
```

```python
data = np.random.randn(5,10)
plt.boxplot(data)
plt.xlabel('Data')
plt.ylabel('value')
plt.title('Box plot')
plt.grid(True)
plt.show()
```

```python
plt.imshow(data, cmap = 'coolwarm', interpolation = 'nearest')
plt.colorbar()
plt.title('Heatmap')
plt.show()
```

```python
categories = ['A','B','C','D']
values1 =[20,12,13,42]
values2 = [15,16,20,10]
plt.bar(categories, values1, color = 'Purple', label = 'data 1')
plt.bar(categories, values2, color = 'orange', bottom = values1, label = 'data')
plt.xlabel('categories')
plt.ylabel('values')
plt.title('Stacked Bar plot')
plt.legend()
plt.show()
```

```python
data = [np.random.normal(0, std, 100) for std in range(1, 4)]
plt.violinplot(data, showmedians=True)
plt.xlabel('Distribution')
plt.ylabel('Value')
plt.title('Violin Plot')
plt.xticks(np.arange(1, len(data) + 1), ['Data 1', 'Data 2', 'Data 3'])
plt.legend(['Median'])
plt.show()
```
